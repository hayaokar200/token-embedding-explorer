## Tokenizer comparison: count tokens and look at the pieces

```python
encoding = tiktoken.get_encoding('o200k_base')

for name, text in texts.items():
    n_words = len(text.split())
    tokens = encoding.encode(text)
    print(f"{name} | words: {n_words} | tokens: {len(tokens)} | tokens per word: {round(len(tokens)/n_words,2)}")
```

```python
arabic_tokens = encoding.encode(texts['arabic'])
english_tokens = encoding.encode(texts['english'])
print(encoding.decode([t]) for t in arabic_tokens)
print(encoding.decode([t]) for t in english_tokens)
```
